## Step 3: change a number and run again
Code can do arithmetic. The cell below asks: if I invest some money and earn the same return every
year, what is it worth later? Change the numbers on the lines marked **←**, then run the cell again.

```
# 🟡 YOUR TURN
INVESTMENT = 1000      # ← try 5000
ANNUAL_RETURN = 0.10   # ← 0.10 means 10% a year; try 0.05 or 0.15
YEARS = 10             # ← try 30

value = INVESTMENT * (1 + ANNUAL_RETURN) ** YEARS
print(f"${INVESTMENT:,.0f} growing at {ANNUAL_RETURN:.0%} a year is worth ${value:,.0f} after {YEARS} years.")
```

## Step 4: look at real data
`mlfin.load_prices("M")` loads month-end prices: one **row** per month, one **column** per stock
(by ticker symbol), plus **SPY**, a fund that tracks the S&P 500 — our stand-in for "the market".
Prices are adjusted for stock splits and dividends (Week 2 explains why).

```
prices = mlfin.load_prices("M")
prices[["AAPL", "MSFT", "JPM", "SPY"]].tail()
```

## Step 5: draw a chart
How much would $1 put into the S&P 500 at the end of 2009 have grown to?

```
growth = prices.loc["2010":, "SPY"] / prices.loc["2009-12-31", "SPY"]
plt.plot(growth.index, growth)
plt.title("Growth of $1 invested in the S&P 500 (SPY) at the end of 2009")
plt.ylabel("Value of $1")
plt.xlabel("Date")
plt.show()
```

*What to look for:* $1 became roughly $8. Notice the sharp drop in early 2020 (COVID) and
the slow slide during 2022 (rising interest rates).

## Step 6: compare a stock you know with the market

```
# 🟡 YOUR TURN
TICKER = "AAPL"   # ← try "NVDA", "KO", "JPM", "XOM", or any ticker you like

stock = prices.loc["2010":, TICKER] / prices.loc["2009-12-31", TICKER]
plt.plot(stock.index, stock, label=TICKER)
plt.plot(growth.index, growth, label="S&P 500 (SPY)")
plt.title(f"Growth of $1: {TICKER} vs the S&P 500")
plt.ylabel("Value of $1")
plt.xlabel("Date")
plt.legend()
plt.show()
```

*What to look for:* is your stock above or below the market line at the end? Some tickers
start later or stop early — the line simply begins or ends where the data does.
If you get a `KeyError`, that ticker isn't in our data (try `"META"` instead of `"FB"`).
